```python
def add_num(x,y):
    return x+y
def sub_num(x,y):
    return x-y
def mul_num(x,y):
    return x*y
def divi_num(x,y):
    return x//y
def modu_div(x,y):
    return x%y
print(add_num(2,6))
print(sub_num(30,3))
print(mul_num(6,6))
print(divi_num(2,12))
print(modu_div(200,98))
```
