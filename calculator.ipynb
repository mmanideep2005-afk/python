```python
def add_num(x,y):
    return x+y
def sub_num(x,y):
    return x-y
def mul_num(x,y):
    return x*y
def divi_num(x,y):
    return x//y
def modu_div(x,y):
    return x%y
print(add_num(2,6))
print(sub_num(30,3))
print(mul_num(6,6))
print(divi_num(2,12))
print(modu_div(200,98))
```

```python
def sub(x,y):
    return x+y
b1=int(input())
b2=int(input())
print(sub(x=b1,y=b2))
print("sum of two numbers done")
def sub2(x,y,z):
    return x+y+z
b1=int(input())
b2=int(input())
b3=int(input())
print(sub2(x=b1,y=b2,z=b3))
print("sum of three numbers done")
def sub1(x,y,c,d):
    return x+y+c+d
b1=int(input())
b2=int(input())
b3=int(input())
b4=int(input())
print(sub1(x=b1,y=b2,c=b3,d=b4))
print("sum of four numbers done")
```
