```python
#dictionary update values and items 
details = {'name': 'manideep', 'batch':20,'course':'python'}
details['course']= 'java'
details['age']= 50
print(details)
```

```python
details = {'name': 'manideep', 'batch':20,'course':'python'}
print(details.keys())
print(details.items())
print(details.values())
```

```python
#pop(key) it returns the values and removes the item from dictionary if key is present, otherwise keys
#popitem() it removes the last item 
details = {'name': 'manideep', 'batch':20,'course':'python',0:1}
print(details.pop(0))
print(details.popitem())
print(details)
```
