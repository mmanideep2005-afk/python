```python
class Employee():
    def __init__(self, emp_id, name):
        self.emp_id = emp_id
        self.name = name
    def calculate_salary(self):
        pass

class SalariedEmployee(Employee):
    def __init__(self, emp_id, name, monthly_salary):
        super().__init__(emp_id, name) 
        self.monthly_salary = monthly_salary
    def calculate_salary(self):
        return self.monthly_salary

class HourlyEmployee(Employee): 
    def __init__(self, emp_id, name, hourly_rate, hours_worked):
        super().__init__(emp_id, name) 
        self.hourly_rate = hourly_rate
        self.hours_worked = hours_worked
    def calculate_salary(self):
        return self.hours_worked * self.hourly_rate

class CommissionEmployee(Employee):
    def __init__(self, emp_id, name, base_salary, commission_rate, total_sales):
        super().__init__(emp_id, name) 
        self.base_salary = base_salary
        self.commission_rate = commission_rate
        self.total_sales = total_sales
    def calculate_salary(self):
        return self.base_salary + (self.commission_rate * self.total_sales)

emp1 = SalariedEmployee(101, "Alice", 5000)
emp2 = HourlyEmployee(102, "Bob", 20, 160)
emp3 = CommissionEmployee(103, "Charlie", 2000, 0.1, 30000)

print(f"Employee: {emp1.name} | Salary: {emp1.calculate_salary()}")
print(f"Employee: {emp2.name} | Salary: {emp2.calculate_salary()}")
print(f"Employee: {emp3.name} | Salary: {emp3.calculate_salary()}")
```
