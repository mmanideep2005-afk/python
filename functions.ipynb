```python
#target element frequency functions
def target_element(arr,n):
    count=0
    for i in arr:
        if i==n:
            count+=1
    return count
def main():
    arr = [12,3,44,9,3,12,3]
    key = 12
    print(target_element(arr,key))
if __name__=="__main__":
    main()
```

```python
#remove the target element from arr
def remove_element(arr,n):
    for i in arr:
        if i==n:
            arr.remove(i)
    return arr
def main():
    arr = [12,3,44,9,3,12,3]
    key = 12
    print(remove_element(arr,key))
if __name__=="__main__":
    main()
```

```python
#square pattern
for i in range(5):
    for j in range(5):
        if  i==0 or j==0 or i==4 or j==4 :  
            print('*',end=' ')
        else:
            print(' ',end=" ")
    print()
```
