```python
class Person():
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def show_details(self):
        return f"My name is {self.name} and iam {self.age} years old"
class Employees(Person):
    def __init__(self,empid,name,job,age):
        super().__init__(name,age)
        self.empid=empid
        self.job=job
p = Person("mani",39)
e = Employees(101,"nani","trainer",90)
print(e.show_details())
```

```python
#method overiding
class P:
    def r(self):
        print("monday")
class C(P):
    def r(self):
        print("tuesday")
e = C()
print(e.r())
```

```python
#Abstraction
from abc import ABC
class Shapes(ABC):
    def area(self):
        pass
    def perimetre(self):
        pass
class Cube(Shapes):
    def area(self,a):
        areaa = 6*a**2
        return areaa
    def perimetre(self,b):
        peri = 12*b
        return peri
e = Cube()
print(e.area(2))
print(e.perimetre(3))
```

```python
import math
from abc import ABC
class shape(ABC):
    def area(self):
        pass
    def perimeter(self):
        pass
class square(shape):
    def __init__(self,side):
        self.side=side
    def area(self):
        return self.side*self.side
    def perimeter(self):
        return 4*self.side
class rectangle(shape):
    def __init__(self,length,breadth):
        self.length=length
        self.breadth=breadth
    def area(self):
        return self.length*self.breadth
    def perimeter(self):
        return 2*(self.length+self.breadth)
class triangle(shape):
    def __init__(self,base,height):
        self.base=base
        self.height=height
    def area(self):
        return 1/2*self.base*self.height
class circle(shape):
    def __init__(self,radius):
        self.radius=radius
    def area(self):
        return math.pi*self.radius**2
    def perimeter(self):
        return 2*math.pi*self.radius
        
a=square(35)
b=rectangle(12,5)
c=triangle(10,5)
d=circle(9)
print("area of the square =",a.area())
print("perimeter of the square =",a.perimeter())
print("area of rectangle",b.area())
print("Perimetre of rectangle",b.perimeter())
print("area of triangle",c.area())
print("area of circle",d.area())
print("Perimetre of circle",d.perimeter())
```
